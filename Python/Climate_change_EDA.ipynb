**Exploratory Data Analysis of Global Climate Change & Extreme Weather Using Python**

**🎯 Problem Statement**

Climate change has become a major global concern, leading to rising temperatures, increased CO₂ emissions, sea level rise, and extreme weather events.

The objective of this project is to perform Exploratory Data Analysis (EDA) on a global climate change dataset to identify trends, analyze relationships among climate indicators, and generate meaningful insights into the impact of climate change across different countries and regions.


**🛠️ Tech Stack- import libraries**

Python – the programming language everything is written in.

Seaborn – built on top of Matplotlib, makes statistical charts (heatmaps, boxplots, distributions) look better with less code.

**1. Import Libraries**

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("global_climate_change_extreme_weather_2026-selected-columns.csv")
```

**4.Data Cleaning**

```python
#Check Missing Values
print("Missing Values in Each Column:\n")
print(df.isnull().sum())
```

```python

df.notnull()
     
```

```python
# ===============================
# Fill Missing Numerical Values
# ===============================

numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_columns:
    df[col].fillna(df[col].median(), inplace=True)

print("Missing values handled successfully.")
```

```python
# ===============================
# Check Duplicate Records
# ===============================

duplicates = df.duplicated().sum()

print("Total Duplicate Records:", duplicates)
```

```python
# ===============================
# Remove Duplicate Records
# ===============================

df.drop_duplicates(inplace=True)

print("Duplicate records removed successfully.")
```

```python
# ===============================
# Check Data Types
# ===============================

print(df.dtypes)
```

```python
# ===============================
# Boxplots for Outlier Detection
# ===============================

numerical_columns = df.select_dtypes(include=['int64','float64']).columns

for col in numerical_columns:
    plt.figure(figsize=(2,1))
    sns.boxplot(x=df[col])

    plt.title(f"Boxplot of {col}")
    plt.show()
```

**5. Univariate Analysis**

This means analyzing one variable at a time.

```python
# ==========================================
# Distribution of Average Temperature
# ==========================================

plt.figure(figsize=(5,3))

sns.histplot(df['avg_temperature_c'], bins=30, kde=True, color='skyblue')

plt.title("Distribution of Average Temperature")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Frequency")

plt.show()
```

```python
# ==========================================
# Distribution of Temperature Change
# ==========================================

plt.figure(figsize=(5,3))
sns.boxplot(x=df["temperature_change_c"])
plt.title("Temperature Change")
plt.show()
```

```python
# ==========================================
# Distribution of CO₂ Emissions
# ==========================================

plt.figure(figsize=(5,3))

sns.histplot(df['co2_emissions_mt'], bins=30, kde=True, color='green')

plt.title("Distribution of CO₂ Emissions")

plt.xlabel("CO₂ Emissions (Metric Tons)")
plt.ylabel("Frequency")

plt.show()
```

```python
# Sea rise level

plt.figure(figsize=(5,3))
sns.boxplot(x=df["sea_level_rise_mm"])
plt.title("Sea Level Rise")
plt.show()
```

```python
#food risk
plt.figure(figsize=(5,3))
sns.countplot(x="flood_risk", data=df)
plt.title("Flood Risk")
plt.show()
```

```python
# Drought Risk → Count Plot
plt.figure(figsize=(5,3))
sns.countplot(x="drought_risk", data=df)
plt.title("Drought Risk")
plt.show()
```

```python
#Heatwave Days
plt.figure(figsize=(4,2))
sns.histplot(df["heatwave_days"], bins=20)
plt.title("Heatwave Days")
plt.show()
```

```python
#Top 10 Countries → Bar Chart
top = df["country"].value_counts().head(10)

plt.figure(figsize=(5,3))
top.plot(kind="bar")

plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

```python
# Regions → Bar Chart
top = df["region"].value_counts().head(10)

plt.figure(figsize=(5,3))
top.plot(kind="bar", color="green")

plt.title("Top 10 Regions")
plt.xlabel("Region")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.show()
```

**6. Bivariate Analysis**

```python
plt.figure(figsize=(5,3))

sns.scatterplot(x="co2_emissions_mt",
                y="avg_temperature_c",
                data=df)

plt.title("Average Temperature vs CO₂ Emissions")
plt.xlabel("CO₂ Emissions")
plt.ylabel("Average Temperature")

plt.show()
```

```python
plt.figure(figsize=(5,3))

sns.scatterplot(x="temperature_change_c",
                y="heatwave_days",
                data=df)

plt.title("Temperature Change vs Heatwave Days")
plt.xlabel("Temperature Change")
plt.ylabel("Heatwave Days")

plt.show()
```

```python
plt.figure(figsize=(5,3))

sns.scatterplot(x="temperature_change_c",
                y="heatwave_days",
                data=df)

plt.title("Temperature Change vs Heatwave Days")
plt.xlabel("Temperature Change")
plt.ylabel("Heatwave Days")

plt.show()
```

```python
top_temp = df.groupby("country")["avg_temperature_c"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,3))

top_temp.plot(kind="bar", color="orange")

plt.title("Top 10 Countries by Average Temperature")
plt.xlabel("Country")
plt.ylabel("Average Temperature")

plt.xticks(rotation=45)

plt.show()
```

```python
top_co2 = df.groupby("country")["co2_emissions_mt"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,3))

top_co2.plot(kind="bar", color="green")

plt.title("Top 10 Countries by CO₂ Emissions")
plt.xlabel("Country")
plt.ylabel("CO₂ Emissions")

plt.xticks(rotation=45)

plt.show()
```

**7. Multivariate Analysis**

```python
plt.figure(figsize=(5,3))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            linewidths=0.5)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.pairplot(df[
    ['avg_temperature_c',
     'temperature_change_c',
     'co2_emissions_mt',
     'sea_level_rise_mm']
])

plt.show()
```

```python
top_temp = df.groupby("country")["avg_temperature_c"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,3))

top_temp.plot(kind="bar", color="tomato")

plt.title("Top 10 Countries with Highest Average Temperature")
plt.xlabel("Country")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=45)

plt.show()
```

```python
top_co2 = df.groupby("country")["co2_emissions_mt"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,3))

top_co2.plot(kind="bar", color="green")

plt.title("Top 10 CO₂ Emitting Countries")
plt.xlabel("Country")
plt.ylabel("CO₂ Emissions")
plt.xticks(rotation=45)

plt.show()
```

```python
top_sea = df.groupby("country")["sea_level_rise_mm"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(5,3))

top_sea.plot(kind="bar", color="skyblue")

plt.title("Top 10 Countries with Highest Sea Level Rise")
plt.xlabel("Country")
plt.ylabel("Sea Level Rise (mm)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(4,2))

sns.boxplot(x="region",
            y="avg_temperature_c",
            data=df)

plt.xticks(rotation=90)

plt.title("Average Temperature by Region")

plt.show()
```
